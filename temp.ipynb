```python
target = "Cash_Demand_Next_Day"

X_train = train.drop(columns=[target, "Date"])
y_train = train[target]

X_val = val.drop(columns=[target, "Date"])
y_val = val[target]

X_test = test.drop(columns=[target, "Date"])
y_test = test[target]

print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("X_test:", X_test.shape)
```

```python
drop_features = [
    "Days_Since_Previous",
    "Previous_Observed_Demand"
]

X_train = X_train.drop(columns=drop_features)
X_val = X_val.drop(columns=drop_features)
X_test = X_test.drop(columns=drop_features)

print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("X_test:", X_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

model_lr = LinearRegression()

model_lr.fit(X_train, y_train)

y_val_pred = model_lr.predict(X_val)

mae = mean_absolute_error(y_val, y_val_pred)
rmse = np.sqrt(mean_squared_error(y_val, y_val_pred))
r2 = r2_score(y_val, y_val_pred)

print("Linear Regression — Validation")
print("MAE :", mae)
print("RMSE:", rmse)
print("R²  :", r2)
```

```python
from sklearn.tree import DecisionTreeRegressor

model_dt = DecisionTreeRegressor(
    random_state=42,
    max_depth=10
)

model_dt.fit(X_train, y_train)

y_val_pred_dt = model_dt.predict(X_val)

mae_dt = mean_absolute_error(y_val, y_val_pred_dt)
rmse_dt = np.sqrt(mean_squared_error(y_val, y_val_pred_dt))
r2_dt = r2_score(y_val, y_val_pred_dt)

print("Decision Tree — Validation")
print("MAE :", mae_dt)
print("RMSE:", rmse_dt)
print("R²  :", r2_dt)
```

```python
from sklearn.ensemble import RandomForestRegressor

model_rf = RandomForestRegressor(
    n_estimators=200,
    max_depth=15,
    random_state=42,
    n_jobs=-1
)

model_rf.fit(X_train, y_train)

y_val_pred_rf = model_rf.predict(X_val)

mae_rf = mean_absolute_error(y_val, y_val_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_val, y_val_pred_rf))
r2_rf = r2_score(y_val, y_val_pred_rf)

print("Random Forest — Validation")
print("MAE :", mae_rf)
print("RMSE:", rmse_rf)
print("R²  :", r2_rf)
```

```python
from sklearn.ensemble import GradientBoostingRegressor

model_gb = GradientBoostingRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)

model_gb.fit(X_train, y_train)

y_val_pred_gb = model_gb.predict(X_val)

mae_gb = mean_absolute_error(y_val, y_val_pred_gb)
rmse_gb = np.sqrt(mean_squared_error(y_val, y_val_pred_gb))
r2_gb = r2_score(y_val, y_val_pred_gb)

print("Gradient Boosting — Validation")
print("MAE :", mae_gb)
print("RMSE:", rmse_gb)
print("R²  :", r2_gb)
```
