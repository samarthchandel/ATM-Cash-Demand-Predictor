```python
import pandas as pd

df = pd.read_csv(
    r"C:\coding samaan\Project\ATM Cash Demand Predictor\atm_cash_management_dataset.csv"
)

print("Dataset shape:", df.shape)
df.head()
```

```python
print("Dataset shape:", df.shape)

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
df["Date"] = pd.to_datetime(df["Date"])

df = df.sort_values(
    ["ATM_ID", "Date", "Time_of_Day"]
).reset_index(drop=True)

print("Date range:", df["Date"].min(), "to", df["Date"].max())
print("Number of ATMs:", df["ATM_ID"].nunique())
```

```python
df_clean = df.drop_duplicates(
    subset=["ATM_ID", "Date", "Time_of_Day"],
    keep="first"
).copy()

print("Original rows:", len(df))
print("Cleaned rows:", len(df_clean))
```

```python
daily_atm = (
    df_clean
    .groupby(["ATM_ID", "Date"], as_index=False)
    .agg({
        "Total_Withdrawals": "mean",
        "Total_Deposits": "mean",
        "Holiday_Flag": "max",
        "Special_Event_Flag": "max",
        "Cash_Demand_Next_Day": "mean"
    })
)

daily_atm = daily_atm.sort_values(
    ["ATM_ID", "Date"]
).reset_index(drop=True)

print("Daily dataset shape:", daily_atm.shape)
print("Number of ATMs:", daily_atm["ATM_ID"].nunique())
print("Unique ATM-days:", daily_atm[["ATM_ID", "Date"]].drop_duplicates().shape[0])

daily_atm.head(10)
```

```python
daily_atm = daily_atm.rename(
    columns={
        "Cash_Demand_Next_Day": "Observed_Demand"
    }
)

daily_atm.head()
daily_atm["Target_Demand"] = (
    daily_atm
    .groupby("ATM_ID")["Observed_Demand"]
    .shift(-1)
)

print(daily_atm.head(10))
print("Rows before removing final observations:", len(daily_atm))
print("Missing targets:", daily_atm["Target_Demand"].isna().sum())
forecast_df = daily_atm.dropna(
    subset=["Target_Demand"]
).copy()

print("Forecasting dataset shape:", forecast_df.shape)


```

```python
forecast_df = forecast_df.sort_values(
    ["ATM_ID", "Date"]
).reset_index(drop=True)

# Previous observed demand
forecast_df["Lag_1_Demand"] = (
    forecast_df
    .groupby("ATM_ID")["Observed_Demand"]
    .shift(1)
)

# Demand from two observations ago
forecast_df["Lag_2_Demand"] = (
    forecast_df
    .groupby("ATM_ID")["Observed_Demand"]
    .shift(2)
)

# Demand from three observations ago
forecast_df["Lag_3_Demand"] = (
    forecast_df
    .groupby("ATM_ID")["Observed_Demand"]
    .shift(3)
)

# Rolling averages using ONLY previous observations
forecast_df["Rolling_3_Mean"] = (
    forecast_df
    .groupby("ATM_ID")["Observed_Demand"]
    .transform(
        lambda x: x.shift(1).rolling(3).mean()
    )
)

forecast_df["Rolling_5_Mean"] = (
    forecast_df
    .groupby("ATM_ID")["Observed_Demand"]
    .transform(
        lambda x: x.shift(1).rolling(5).mean()
    )
)

print(forecast_df.head(10))
```

```python
print("Total forecasting rows:", len(forecast_df))

print("\nMissing values:")
print(
    forecast_df[
        [
            "Lag_1_Demand",
            "Lag_2_Demand",
            "Lag_3_Demand",
            "Rolling_3_Mean",
            "Rolling_5_Mean"
        ]
    ].isnull().sum()
)
```

```python
baseline_df = forecast_df.dropna(
    subset=["Lag_1_Demand"]
).copy()

print("Baseline dataset shape:", baseline_df.shape)
print("Missing values:")
print(
    baseline_df[
        [
            "Lag_1_Demand",
            "Total_Withdrawals",
            "Total_Deposits",
            "Holiday_Flag",
            "Special_Event_Flag",
            "Target_Demand"
        ]
    ].isnull().sum()
)
```

```python
print("Start date:", baseline_df["Date"].min())
print("End date:", baseline_df["Date"].max())

print("\nDate quantiles:")
print(
    baseline_df["Date"].quantile([0.70, 0.85])
)
```

```python
train = baseline_df[
    baseline_df["Date"] < "2023-05-24"
].copy()

val = baseline_df[
    (baseline_df["Date"] >= "2023-05-24") &
    (baseline_df["Date"] < "2023-09-13")
].copy()

test = baseline_df[
    baseline_df["Date"] >= "2023-09-13"
].copy()

print("Train:", train.shape)
print("Validation:", val.shape)
print("Test:", test.shape)

print("\nTrain dates:", train["Date"].min(), "to", train["Date"].max())
print("Validation dates:", val["Date"].min(), "to", val["Date"].max())
print("Test dates:", test["Date"].min(), "to", test["Date"].max())
```

```python
features = [
    "Lag_1_Demand",
    "Total_Withdrawals",
    "Total_Deposits",
    "Holiday_Flag",
    "Special_Event_Flag"
]

target = "Target_Demand"

X_train = train[features]
y_train = train[target]

X_val = val[features]
y_val = val[target]

X_test = test[features]
y_test = test[target]

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_val:", X_val.shape)
print("y_val:", y_val.shape)

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
model_lr = LinearRegression()

model_lr.fit(X_train, y_train)

y_val_pred = model_lr.predict(X_val)

mae = mean_absolute_error(y_val, y_val_pred)
rmse = np.sqrt(mean_squared_error(y_val, y_val_pred))
r2 = r2_score(y_val, y_val_pred)

print("Linear Regression — Validation Results")
print("---------------------------------------")
print(f"MAE  : ₹{mae:,.2f}")
print(f"RMSE : ₹{rmse:,.2f}")
print(f"R²   : {r2:.4f}")
```

```python
forecast_df["Days_To_Next"] = (
    forecast_df
    .groupby("ATM_ID")["Date"]
    .diff()
    .shift(-1)
    .dt.days
)

print(forecast_df["Days_To_Next"].describe())

print("\nDays to next observation:")
print(
    forecast_df["Days_To_Next"]
    .value_counts()
    .sort_index()
    .head(20)
)
```

```python
model_df = df_clean.copy()

features = [
    "Total_Withdrawals",
    "Total_Deposits",
    "Holiday_Flag",
    "Special_Event_Flag"
]

target = "Cash_Demand_Next_Day"

print("Model dataset:", model_df.shape)
print(model_df[features + [target]].head())
```

```python
train = model_df[
    model_df["Date"] < "2023-05-28"
].copy()

val = model_df[
    (model_df["Date"] >= "2023-05-28") &
    (model_df["Date"] < "2023-09-18")
].copy()

test = model_df[
    model_df["Date"] >= "2023-09-18"
].copy()

print("Train:", train.shape)
print("Validation:", val.shape)
print("Test:", test.shape)
```

```python
X_train = train[features]
y_train = train[target]

X_val = val[features]
y_val = val[target]

X_test = test[features]
y_test = test[target]

print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("X_test:", X_test.shape)
```

```python
model_lr = LinearRegression()

model_lr.fit(X_train, y_train)

y_val_pred = model_lr.predict(X_val)

print("Model trained successfully.")
```

```python
mae = mean_absolute_error(y_val, y_val_pred)
rmse = np.sqrt(mean_squared_error(y_val, y_val_pred))
r2 = r2_score(y_val, y_val_pred)

print("Validation MAE :", mae)
print("Validation RMSE:", rmse)
print("Validation R²  :", r2)
```

```python
coefficients = pd.DataFrame({
    "Feature": features,
    "Coefficient": model_lr.coef_
})

print(coefficients)
print("Intercept:", model_lr.intercept_)
```

```python
model_lr_withdrawals = LinearRegression()

model_lr_withdrawals.fit(
    X_train[["Total_Withdrawals"]],
    y_train
)

y_val_pred_withdrawals = model_lr_withdrawals.predict(
    X_val[["Total_Withdrawals"]]
)

mae_withdrawals = mean_absolute_error(
    y_val, y_val_pred_withdrawals
)

rmse_withdrawals = np.sqrt(
    mean_squared_error(y_val, y_val_pred_withdrawals)
)

r2_withdrawals = r2_score(
    y_val, y_val_pred_withdrawals
)

print("Withdrawal-only baseline")
print("Validation MAE :", mae_withdrawals)
print("Validation RMSE:", rmse_withdrawals)
print("Validation R²  :", r2_withdrawals)
```

```python
model_lr_transactions = LinearRegression()

transaction_features = [
    "Total_Withdrawals",
    "Total_Deposits"
]

model_lr_transactions.fit(
    X_train[transaction_features],
    y_train
)

y_val_pred_transactions = model_lr_transactions.predict(
    X_val[transaction_features]
)

mae_transactions = mean_absolute_error(
    y_val, y_val_pred_transactions
)

rmse_transactions = np.sqrt(
    mean_squared_error(y_val, y_val_pred_transactions)
)

r2_transactions = r2_score(
    y_val, y_val_pred_transactions
)

print("Transaction-only model")
print("Validation MAE :", mae_transactions)
print("Validation RMSE:", rmse_transactions)
print("Validation R²  :", r2_transactions)
```

```python
model_gb = GradientBoostingRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)

model_gb.fit(X_train, y_train)

y_val_pred_gb = model_gb.predict(X_val)

mae_gb = mean_absolute_error(y_val, y_val_pred_gb)
rmse_gb = np.sqrt(mean_squared_error(y_val, y_val_pred_gb))
r2_gb = r2_score(y_val, y_val_pred_gb)

print("Gradient Boosting")
print("Validation MAE :", mae_gb)
print("Validation RMSE:", rmse_gb)
print("Validation R²  :", r2_gb)
```

```python
model_rf = RandomForestRegressor(
    n_estimators=200,
    max_depth=15,
    random_state=42,
    n_jobs=-1
)

model_rf.fit(X_train, y_train)

y_val_pred_rf = model_rf.predict(X_val)

mae_rf = mean_absolute_error(y_val, y_val_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_val, y_val_pred_rf))
r2_rf = r2_score(y_val, y_val_pred_rf)

print("Random Forest")
print("Validation MAE :", mae_rf)
print("Validation RMSE:", rmse_rf)
print("Validation R²  :", r2_rf)
```

```python
residuals = y_val - y_val_pred

error_analysis = pd.DataFrame({
    "Actual": y_val.values,
    "Predicted": y_val_pred,
    "Residual": residuals.values
})

print(error_analysis.head())
print("\nResidual statistics:")
print(error_analysis["Residual"].describe())
```

```python
error_analysis["Demand_Level"] = pd.cut(
    error_analysis["Actual"],
    bins=[0, 30000, 45000, 60000, float("inf")],
    labels=["Low", "Medium", "High", "Very High"]
)

grouped_errors = error_analysis.groupby(
    "Demand_Level",
    observed=True
)["Residual"].agg(
    ["count", "mean", "std"]
)

print(grouped_errors)
```

```python
print(
    error_analysis[
        ["Actual", "Predicted", "Residual"]
    ].corr()
)
```
